```python
# **힌트: 폴더 안 파일 목록 가져오기**
import glob

# data/baemin 폴더에서 baemin으로 시작하는 csv 파일 경로를 모두 찾아 리스트로 돌려줌
# * 는 "아무 글자나"라는 뜻 → baemin20261001.csv, baemin20261002.csv ... 전부 찾음
# sorted() 로 날짜 순서대로 정렬
baemin_files = sorted(glob.glob("data/baemin/baemin*.csv"))
baemin_files
```

```python
pd.read_csv(baemin_files[0])
```

```python
coupang_files = sorted(glob.glob("data/coupang/coupang*.csv"))
```

```python
print(len(baemin_files), len(coupang_files))
```

```python
baemin_list = []

for file in baemin_files:
    df = pd.read_csv(file)
    baemin_list.append(df)

baemin = pd.concat(baemin_list, ignore_index=True)
```

```python
baemin = baemin.rename(columns={"order_at": "주문일시", "total_bill":"전체금액","ad_yn":"광고유입여부","cpc":"광고집행비"})
```

```python
baemin["플랫폼"] = "배민"
```

```python
baemin.head()
```

```python
coupang_list = []

for file in coupang_files:
    df = pd.read_csv(file)
    coupang_list.append(df)

coupang = pd.concat(coupang_list, ignore_index=True)
```

```python
coupang = coupang.rename(columns={
    "order_time": "주문일시",
    "total_price":"전체금액",
    "advertisement_yn":"광고유입여부",
})

coupang["광고집행비"] = coupang["전체금액"] * coupang["ad_pct"] / 100
coupang["플랫폼"] = "쿠팡"
```

```python
coupang.head()
```

```python
columns = ["주문일시", "전체금액", "광고유입여부", "플랫폼","광고집행비"]

baemin = baemin[columns]
coupang = coupang[columns]
```

```python
sales = pd.concat([baemin, coupang], ignore_index=True)
sales.sample(5)
```

```python
sales.info()
```

```python
sales["주문일자"] = pd.to_datetime(sales["주문일시"]).dt.date
sales = sales.sort_values(by="주문일자")
sales.head(3)
```

```python
daily_sales = sales.pivot_table(
    index="주문일자",
    columns="플랫폼",
    values="전체금액",
    aggfunc="sum",
    fill_value=0
).reset_index()

daily_sales["합계"] = daily_sales[["배민","쿠팡"]].sum(axis=1)
```

```python
ad_ratio = sales.groupby("플랫폼")["광고유입여부"].value_counts(normalize=True).unstack().fillna(0)
```

```python
ad_ratio = sales.groupby("플랫폼")["광고유입여부"].apply(
    lambda x: (x == "Y").mean() * 100
)
ad_cost = sales.groupby("플랫폼")["광고집행비"].sum()
```

```python
bank = pd.read_csv("data/bank_20261001_20261031.csv",
                   encoding="cp949")
```

```python
food = bank[bank["거래처"] == "식자재마트"]

food["거래일자"] = pd.to_datetime(food["거래일시"]).dt.date
```

```python
food_daily = food.groupby("거래일자")["출금액"].sum().reset_index()

food_daily = food_daily.rename(columns={"거래일자":"주문일자","출금액":"식자재비"})
```

```python
result = daily_sales.merge(food_daily, on="주문일자", how="left")
```

```python
result["식자재비"] = result["식자재비"].fillna(0)
result["매출-식자재비"] = result["합계"] - result["식자재비"]

result = result[["주문일자","배민","쿠팡","합계","식자재비","매출-식자재비"]]
```
