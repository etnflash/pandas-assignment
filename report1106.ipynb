### 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.


```python
mask = (titanic['sex'] == 'male') & (titanic['age'].between(20,40)) & ((titanic['pclass'].isin([1,2])))
m_titanic = titanic[mask][['pclass', 'age', 'fare', 'survived']]
m_titanic
```

```python
print(m_titanic.shape[0])
print(round(m_titanic['survived'].value_counts(normalize=True)[1], 2))
print(m_titanic.sort_values(by='fare', ascending=False).head(3))
```

```python
# 3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
# 4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.
# titanic['sex'] == 'male') & (titanic['age'].between(20,40)) & ((titanic['pclass'].isin([1,2])
titanic.query('sex == "male" & age.between(20,40) & pclass.isin([1,2])').shape[0]
```

```python
print(round(titanic.query('sex == "male" & age.between(20,40) & pclass.isin([1,2])')['survived'].mean(),2))
```

```python
titanic.query('sex == "male" & age.between(20,40) & pclass == 3').shape[0]
```

```python
print(round(titanic.query('sex == "male" & age.between(20,40) & pclass == 3')['survived'].mean(), 2))
```

1·2등석 남성 승객의 생존률은 24%, 3등석 남성 승객의 생존률은 15%이다. 따라서 같은 나이 범위의 남성이라도 1·2등석 승객의 생존률이 3등석보다 9% 높다.
